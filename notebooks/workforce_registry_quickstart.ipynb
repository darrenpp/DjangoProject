# NDOH Workforce Registry Quickstart

Basic notebook to connect to Django ORM and inspect core workforce metrics.

```python
import os
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))

os.environ.setdefault("DJANGO_SETTINGS_MODULE", "ndoh_workforce_registry.settings")

import django
django.setup()

print("Django initialized")
```

```python
from apps.workforce.models import (
    MedicalDoctor,
    NursingProfessional,
    Midwife,
    CommunityHealthWorker,
    HealthStudent,
    Facility,
    TrainingInstitution,
)

summary = {
    "medical_count": MedicalDoctor.objects.count(),
    "nursing_count": NursingProfessional.objects.count(),
    "midwife_count": Midwife.objects.count(),
    "chw_count": CommunityHealthWorker.objects.count(),
    "student_count": HealthStudent.objects.count(),
    "facility_count": Facility.objects.count(),
    "institution_count": TrainingInstitution.objects.count(),
}

summary
```

```python
import pandas as pd

df = pd.DataFrame([summary])
df
```
